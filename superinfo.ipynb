```python
import numpy as np
#module to count occurences of elements in sequences.
from collections import Counter
```

```python

def calculate_superinformation(sequence, symbols_per_block, num_bins):
    #Divide the sequence into blocks, by slicing the sequences into number of blocks required.
    blocks = [sequence[i:i+symbols_per_block] for i in range(0, len(sequence), symbols_per_block)]

    #Calculate entropy for each block
    def calculate_entropy(block):
        counts = Counter(block)
        total = len(block)
        entropy = sum(-(count / total) * np.log2(count / total) for count in counts.values() if count != 0)
        return entropy

    entropies = [calculate_entropy(block) for block in blocks]

    #Construct histogram of entropies
    histogram = np.histogram(entropies, bins=num_bins, range=(0, 1))

    #Normalize the histogram
    probabilities = histogram[0] / len(entropies)

    #Calculate superinformation
    superinformation = -np.sum(probabilities * np.log2(np.clip(probabilities, 1e-10, None)))

    return superinformation
```

```python
# Example usage
sequence = "ATCGATCGATCGAAAATCTCGGGTTTTAAA"
symbols_per_block = 4
num_bins = 20

superinformation = calculate_superinformation(sequence, symbols_per_block, num_bins)
print("Superinformation:", superinformation)
```
